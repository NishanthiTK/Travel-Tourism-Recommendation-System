# Travel & Tourism Recommendation System

## 1. Project Introduction

The Travel & Tourism Recommendation System is a Machine Learning-based application that recommends suitable travel destinations based on user preferences.

## 2. Business Problem

Travelers often find it difficult to select destinations that match their budget, interests, travel type and preferences.

This project aims to solve this problem through an intelligent recommendation system.

## 3. Project Objectives

- Recommend suitable travel destinations.
- Analyze destination information and customer preferences.
- Clean and preprocess the dataset.
- Build a Machine Learning recommendation algorithm.
- Deploy the recommendation system using Streamlit.

## 4. Technologies Used

- Python
- NumPy
- Pandas
- Matplotlib
- Seaborn
- Scikit-learn
- TF-IDF
- Cosine Similarity
- Streamlit

## 5. Machine Learning Approach
Used both Supervised and Unsupervised Machine Learning. It is Hybrid Machine Learning

                         
Content-Based Recommendation	-      TF-IDF + Cosine Similarity	          ->      Unsupervised / Similarity-based

Rating Prediction	            -      Random Forest Regression	              ->     Supervised

Rating Classification	        -      Logistic Regression	                  ->      Supervised



## 6. Expected Output

An interactive web application that recommends the top 5 travel destinations based on user preferences.

```python
ml_features = [
    'Name',
    'State',
    'City',
    'Zone',
    'Type',
    'Significance',
    'Best Time to visit'
]

ml_data = clean_data[ml_features].copy()

print("ML Feature Dataset Created!")
print("Shape:", ml_data.shape)

ml_data.head()
```

```python
ml_data['Combined_Features'] = (
    ml_data['State'] + ' ' +
    ml_data['City'] + ' ' +
    ml_data['Zone'] + ' ' +
    ml_data['Type'] + ' ' +
    ml_data['Significance'] + ' ' +
    ml_data['Best Time to visit']
)

print("Combined Features Created!")

ml_data[['Name', 'Combined_Features']].head()
```

```python
print("Total Records:", len(ml_data))

print(
    "Missing Combined Features:",
    ml_data['Combined_Features'].isnull().sum()
)

print(
    "Empty Combined Features:",
    (ml_data['Combined_Features'].str.strip() == '').sum()
)

print(
    "Duplicate Combined Features:",
    ml_data['Combined_Features'].duplicated().sum()
)
```

```python
duplicate_features = ml_data[
    ml_data['Combined_Features'].duplicated(keep=False)
].sort_values('Combined_Features')

print("Duplicate Feature Records:")

duplicate_features[
    ['Name', 'State', 'City', 'Type', 'Significance', 'Combined_Features']
]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF
tfidf = TfidfVectorizer()

# Convert combined text features into numerical vectors
tfidf_matrix = tfidf.fit_transform(ml_data['Combined_Features'])

print("TF-IDF Conversion Completed!")
print("Matrix Shape:", tfidf_matrix.shape)
print("Number of Attractions:", len(ml_data))
print("Number of Unique Words:", len(tfidf.get_feature_names_out()))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between all tourist attractions
cosine_sim = cosine_similarity(tfidf_matrix)

print("Cosine Similarity Calculated!")
print("Similarity Matrix Shape:", cosine_sim.shape)

# Display similarity scores for first 5 attractions
print("\nSimilarity Scores:")
print(cosine_sim[:5, :5])
```

```python
def recommend_places(place_name, top_n=5):

    # Find the selected attraction
    place_index = ml_data[
        ml_data['Name'].str.lower() == place_name.lower()
    ].index

    if len(place_index) == 0:
        return "Attraction not found!"

    place_index = place_index[0]

    # Get similarity scores
    similarity_scores = list(enumerate(cosine_sim[place_index]))

    # Sort by similarity score (highest first)
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Exclude the selected attraction itself
    similarity_scores = similarity_scores[1:top_n+1]

    # Get recommended attraction details
    recommended_indices = [i[0] for i in similarity_scores]

    recommendations = ml_data.iloc[recommended_indices][
        ['Name', 'State', 'City', 'Type', 'Significance']
    ].copy()

    recommendations['Similarity Score'] = [
        round(score, 3) for _, score in similarity_scores
    ]

    return recommendations
```

```python
# Prepare target variables

# Target for Linear Regression
y_regression = clean_data['Google review rating']

# Target for Logistic Regression
clean_data['High_Rating'] = (
    clean_data['Google review rating'] >= 4.5
).astype(int)

y_classification = clean_data['High_Rating']

# Check classification distribution
print("Rating Classification:")
print(y_classification.value_counts())

print("\nPercentage Distribution:")
print(
    (y_classification.value_counts(normalize=True) * 100).round(2)
)
```

```python
from sklearn.model_selection import train_test_split

# Select input features
X = clean_data[
    [
        'Zone',
        'State',
        'City',
        'Type',
        'Significance',
        'Best Time to visit',
        'Entrance Fee in INR',
        'Time needed to visit in hrs',
        'Airport with 50km Radius',
        'DSLR Allowed'
    ]
]

# Target variables
y_regression = clean_data['Google review rating']
y_classification = clean_data['High_Rating']

print("Input Features Shape:", X.shape)
print("Regression Target Shape:", y_regression.shape)
print("Classification Target Shape:", y_classification.shape)
```

```python
# Split data for Linear Regression
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X, y_regression,
    test_size=0.2,
    random_state=42
)

# Split data for Logistic Regression
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X, y_classification,
    test_size=0.2,
    random_state=42,
    stratify=y_classification
)

print("LINEAR REGRESSION")
print("Training Features:", X_train_reg.shape)
print("Testing Features:", X_test_reg.shape)

print("\nLOGISTIC REGRESSION")
print("Training Features:", X_train_clf.shape)
print("Testing Features:", X_test_clf.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical and numerical columns
categorical_features = X.select_dtypes(
    include=['object']
).columns.tolist()

numerical_features = X.select_dtypes(
    include=['number']
).columns.tolist()

# Create preprocessing transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical_features),
        ('numerical', 'passthrough', numerical_features)
    ]
)

print("Categorical Features:", categorical_features)
print("\nNumerical Features:", numerical_features)
print("\nPreprocessor created successfully!")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# Create Linear Regression pipeline
linear_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

print("Linear Regression Pipeline Created Successfully!")
```

```python
# Train the Linear Regression model
linear_model.fit(X_train_reg, y_train_reg)

print("Linear Regression Model Trained Successfully!")

# Predict ratings for test data
y_pred_reg = linear_model.predict(X_test_reg)

print("Predictions Generated Successfully!")
print("First 10 Actual Ratings:", y_test_reg.head(10).values)
print("First 10 Predicted Ratings:", y_pred_reg[:10].round(2))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mae = mean_absolute_error(y_test_reg, y_pred_reg)
mse = mean_squared_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_reg, y_pred_reg)

print("LINEAR REGRESSION - MODEL EVALUATION")
print("------------------------------------")
print("Mean Absolute Error (MAE):", round(mae, 3))
print("Mean Squared Error (MSE):", round(mse, 3))
print("Root Mean Squared Error (RMSE):", round(rmse, 3))
print("R2 Score:", round(r2, 3))
print("R2 Score (%):", round(r2 * 100, 2), "%")
```

```python
from sklearn.ensemble import RandomForestRegressor
# Create Random Forest Regression pipeline
rf_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])

# Train model
rf_model.fit(X_train_reg, y_train_reg)

# Generate predictions
rf_pred = rf_model.predict(X_test_reg)

# Evaluate model
mae_rf = mean_absolute_error(y_test_reg, rf_pred)
mse_rf = mean_squared_error(y_test_reg, rf_pred)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test_reg, rf_pred)

print("RANDOM FOREST REGRESSION")
print("------------------------")
print("MAE:", round(mae_rf, 3))
print("MSE:", round(mse_rf, 3))
print("RMSE:", round(rmse_rf, 3))
print("R2 Score:", round(r2_rf, 3))
```
